```python
import requests
import json
import time

# إعداد البيانات الأساسية
api_base_url = "https://www.universal-tutorial.com/api/"
api_token = "qeutI6-_wvovNmJrB_fuH8ocRN1H7VW8vOkTyx2UwbT3ukGaHwyTPVwCyJQKEhpDwwY"
user_email = "engmoustafa02@gmail.com"
access_token = None

def get_access_token():
    global access_token
    url = api_base_url + "getaccesstoken"
    headers = {
        "Accept": "application/json",
        "api-token": api_token,
        "user-email": user_email
    }
    response = requests.get(url, headers=headers)
    if response.status_code == 200:
        access_token = response.json().get("auth_token")
    else:
        print("Failed to obtain access token. Exiting...")
        exit()

def get_countries():
    url = api_base_url + "countries/"
    headers = {
        "Authorization": f"Bearer {access_token}",
        "Accept": "application/json"
    }
    response = requests.get(url, headers=headers)
    if response.status_code == 200:
        return response.json()
    else:
        print("Failed to retrieve countries data. Exiting...")
        exit()

def get_states(country_name):
    url = api_base_url + f"states/{country_name}"
    headers = {
        "Authorization": f"Bearer {access_token}",
        "Accept": "application/json"
    }
    response = requests.get(url, headers=headers)
    if response.status_code == 200:
        return response.json()
    else:
        print(f"Failed to retrieve states data for {country_name}. Exiting...")
        exit()

def get_cities(state_name):
    url = api_base_url + f"cities/{state_name}"
    headers = {
        "Authorization": f"Bearer {access_token}",
        "Accept": "application/json"
    }
    response = requests.get(url, headers=headers)
    if response.status_code == 200:
        return response.json()
    else:
        print(f"Failed to retrieve cities data for {state_name}. Exiting...")
        exit()

def create_cities_json():
    global access_token
    get_access_token()
    if not access_token:
        print("No access token available. Exiting...")
        exit()

    data = {}
    countries = get_countries()

    if isinstance(countries, list):
        for country in countries:
            country_name = country.get('country_name')
            if country_name:
                data[country_name] = {}
                states = get_states(country_name)
                if isinstance(states, list):
                    for state in states:
                        state_name = state.get('state_name')
                        if state_name:
                            cities = get_cities(state_name)
                            if isinstance(cities, list):
                                data[country_name][state_name] = [city.get('city_name') for city in cities]
                            else:
                                print(f"Failed to retrieve cities data for {state_name}. Skipping...")
                            time.sleep(1)
                else:
                    print(f"Failed to retrieve states data for {country_name}. Skipping...")

    with open('cities_data.json', 'w', encoding='utf-8') as f:
        json.dump(data, f, ensure_ascii=False, indent=4)
    print("Cities data successfully saved to cities_data.json.")

create_cities_json()
```
